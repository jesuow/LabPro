3. Diferencia entre transformaciones y acciones en Spark. Proporciona ejemplos de cada una. ¿Qué ocurre internamente cuando se ejecuta una acción?
4. Explica la importancia del particionamiento en Spark y cómo afecta el rendimiento del procesamiento de datos.

**Respuestas**

1- Apache Spark es un motor unificado que se utiliza para el procesamiento de grandes volúmenes de datos. Esta se centra en cuatro pilares en su diseño, la velocidad, facilidad de uso, mosularidad y extensibilidad. Esto le permite ser una solución flexible y avanzada para el procesamiento de grandes volúmenes de datos. Sus ventajas sobre Pandas incluyen el procesamiento en memoria para mayor velocidad y la capacidad de trabajar en clústeres distribuidos para manejar grandes conjuntos de datos.

2-  Un RDD (Resilient Distributed Dataset) es una colección distribuida e inmutable de datos que permite operaciones en paralelo y es fundamental para Spark. Una característica principal es su tolerancia a fallos mediante la replicación de datos. Los DataFrames son una abstracción superior a los RDDs que proporcionan una API optimizada y permiten consultas SQL, aunque internamente Spark sigue utilizando RDDs para el procesamiento.

3- Las transformaciones (e.g., groupby(), select()) crean un nuevo DataFrame a partir del original sin ejecutar ninguna operación, mientras que las acciones (e.g., count(), show()) disparan la ejecución de las transformaciones acumuladas. Al ejecutar una acción, Spark construye un plan de ejecución, que se descompone en stages y tasks, y se ejecuta el DAG para obtener los resultados.

4- El particionamiento en Spark divide los datos en partes más pequeñas llamadas particiones, que se distribuyen entre los nodos del clúster. Esto es crucial porque permite que cada nodo procese una partición de datos simultáneamente, aumentando la velocidad del procesamiento al aprovechar el paralelismo. Además, reduce el tiempo necesario para mover datos entre nodos, optimizando el uso de recursos y mejorando la eficiencia general del procesamiento.

5- El Spark Driver gestiona la ejecución de la aplicación, coordina la distribución de tareas y supervisa el progreso general, mientras que los Spark Executors son responsables de ejecutar las tareas en paralelo sobre particiones de datos, realizando cálculos y almacenando los resultados. El Driver divide la aplicación en trabajos y etapas, distribuyendo las tareas a los Executors, que procesan datos en paralelo y almacenan resultados intermedios, lo que optimiza el rendimiento y la eficiencia del procesamiento en el clúster.

6- El Catalyst Optimizer en Apache Spark es una herramienta que optimiza las consultas SQL para mejorar el rendimiento. Analiza la consulta creando un Árbol de Sintaxis Abstracta (AST), genera y selecciona planes de ejecución lógicos, escoge los operadores físicos más adecuados y produce código optimizado en Java. Su función principal es transformar las consultas en planes eficientes para acelerar el procesamiento de datos en el clúster.

**¿Cuál/es columna/s tiene/n un tipo de dato que no es el adecuado y por qué?**

````python
#Cree una función cast_columns que permita cambiar el tipo de datos de las columnas problemáticas.
#Luego utilice esta función respecto a lo respondido en la pregunta anterior.
def cast_columns(df, cols_types):
    """
    Cambia el tipo de múltiples columnas en un DataFrame de Spark.

    Parámetros:
    - df: DataFrame de Spark.
    - cols_types: Diccionario con nombres de columnas como claves y tipos de datos de Spark como valores.

    Retorna:
    - DataFrame de Spark con tipos de columnas modificados.
    """
    for col_name, col_type in cols_types.items():
        df = df.withColumn(col_name, df[col_name].cast(col_type))
    return df
    pass
````

2.   Crear una columna llamada "Transaction Month" con el mes en que se realiza una transacción.

````python
# 1  Agregar una columna llamada "Transaction bp" con el
#**monto total** de la transacción en bodoque pesos.
#Se considera que $x$ dólares equivalen a $log(48+|x^{36}|)$ bodoque pesos.
from pyspark.sql.functions import log, abs, col

# Agregar columna "Transaction bp" con la fórmula log(48 + |x36|)
df = df.withColumn("Transaction bp", log(48 + abs(col("Transaction Amount") * 36)))

df.show(5)
````

````python
## 2 Crear una columna llamada "Transaction Month" con el mes en que se realiza una transacción.
from pyspark.sql.functions import month

# Agregar columna "Transaction Month" con el mes de la transacción
df = df.withColumn("Transaction Month", month(col("Transaction Date")))

df.show(5)
````

````python
# 3 Crear la variable Type of purchase según la catidad de unidades vendidas de acuerdo a
# las siguientes categorías. Compra minorista: 4 productos o menos  Compra mayorista: 5 produtos o más.
df = df.withColumn("Type of purchase", when(col("Quantity") <= 4, "Compra minorista").otherwise("Compra mayorista"))

df.show(5)
````

````python
# 4 Imprima los registros de compras hechas por alienígenas en el comecio mayorista.
df_alien = df.filter((df['Is Fraudulent'] == 1) & (df['Type of purchase'] == 'Compra mayorista')) # Added parentheses around conditions
df_alien.show(5)

df_alien.count()
````

````python
#Cuente la cantidad de compras realizadas por humanos
#y la cantidad de compras realizadas por alienígenas. Utilice .groupby().
df.groupBy('Is Fraudulent').agg({'Quantity':'count'}).show()
````

````python
#Muestre una tabla con la recaudación promedio por transacción para
#cada método de pago, tanto para humanos como alienígenas. Utilice pivot().
df.groupBy('Payment Method').pivot('Is Fraudulent').agg({'Transaction Amount':'mean'}).show()
````

####Visualización de variables categóricas

````python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
````

````python
#Dividimos el dataset en tipo humano o alien
humanos = pandas_df[pandas_df['Is Fraudulent'] == 0]
aliens = pandas_df[pandas_df['Is Fraudulent'] == 1]
````

````python
print(humanos.shape)
print(aliens.shape)
````

````python
# Creamos subplots
fig = make_subplots(rows=1, cols=3, subplot_titles=("Device Used", "Product Category", "Payment Method"))

# Variables categóricas para cada grupo
device_used = pandas_df['Device Used'].unique()
product_category = pandas_df['Product Category'].unique()
payment_methods = pandas_df['Payment Method'].unique()

# Gráfico 1: Dispositivo usado por tipo de cliente
humanos_device_used = humanos['Device Used'].value_counts().reindex(device_used, fill_value=0)
aliens_device_used = aliens['Device Used'].value_counts().reindex(device_used, fill_value=0)

fig.add_trace(
    go.Bar(name="Humanos", x=device_used, y=humanos_device_used, marker_color='purple', showlegend=True),
    row=1, col=1
)
fig.add_trace(
    go.Bar(name="Aliens", x=device_used, y=aliens_device_used, marker_color='orange', showlegend=True),
    row=1, col=1
)

# Gráfico 2: Categoría de producto por tipo de cliente
humanos_product_category = humanos['Product Category'].value_counts().reindex(product_category, fill_value=0)
aliens_product_category = aliens['Product Category'].value_counts().reindex(product_category, fill_value=0)

fig.add_trace(
    go.Bar(name="Humanos", x=product_category, y=humanos_product_category, marker_color='purple', showlegend=False),
    row=1, col=2
)
fig.add_trace(
    go.Bar(name="Aliens", x=product_category, y=aliens_product_category, marker_color='orange', showlegend=False),
    row=1, col=2
)

# Gráfico 3: Método de pago por tipo de cliente
humanos_payment_method = humanos['Payment Method'].value_counts().reindex(payment_methods, fill_value=0)
aliens_payment_method = aliens['Payment Method'].value_counts().reindex(payment_methods, fill_value=0)

fig.add_trace(
    go.Bar(name="Humanos", x=payment_methods, y=humanos_payment_method, marker_color='purple', showlegend=False),
    row=1, col=3
)
fig.add_trace(
    go.Bar(name="Aliens", x=payment_methods, y=aliens_payment_method, marker_color='orange', showlegend=False),
    row=1, col=3
)

# Actualizamos el layout del gráfico
fig.update_layout(barmode='group', title="Comparación de Humanos vs Aliens", showlegend=True)

# Mostramos el gráfico
fig.show()
````

En el primer gráfico se puede observar que los humanos utilizan de manera bastante uniforme todos los dispositivos (tablet, desktop y mobile), mientras que los aliens tienen una proporción menor de uso de dispositivos. Esto indica que los humanos dominan el uso de cada dispositivo en las transacciones, y que los aliens, aunque presentes, realizan un número mucho menor de transacciones en comparación con los humanos.

Similar al gráfico anterior, en el gráfico de "Product category" los humanos realizan la mayoría de las transacciones en todas las categorías. En cada categoría, las transacciones de los aliens son muy inferiores, lo que sugiere que los humanos representan la gran mayoría de las compras en todas las categorías de productos. No hay una categoría en la que los aliens destaquen significativamente.

En cuanto a los métodos de pago, la tendencia se mantiene. Los humanos utilizan todos los métodos de pago de manera uniforme (bank transfer, debit card, PayPal, credit card), con un volumen mucho mayor de transacciones que los aliens. Los aliens también utilizan estos métodos, pero en una proporción mucho menor. No hay un método de pago en particular donde los aliens sobresalgan o prefieran más que los humanos.

El equipo de Bodoque e-shop ha solicitado que los datos estén disponibles en una tabla SQL consultable. Además, están interesados en aprovechar las funciones de ventana en SQL para análisis avanzados. Las funciones de ventana permiten realizar cálculos sobre un conjunto de filas que están relacionadas con la fila actual. Por ejemplo, UNBOUNDED PRECEDING se usa para indicar que el rango de la función de ventana comienza desde la primera fila de la partición o del conjunto de resultados, lo cual es útil para calcular sumas acumulativas hasta la fila actual. Las variaciones comunes de este uso incluyen:

- `UNBOUNDED PRECEDING` to `CURRENT ROW`: Calcula desde el inicio de la partición hasta la fila actual.
- `UNBOUNDED PRECEDING` to `UNBOUNDED FOLLOWING`: Cubre todas las filas dentro de la partición.
- `VALUE PRECEDING` to `VALUE FOLLOWING`: Establece un rango específico basado en valores antes y después de la fila actual.

Ejemplo de uso en SQL:

```sql
STAT(COL1_NAME) OVER (PARTITION BY COL2_NAME ORDER BY COL3_NAME ROWS BETWEEN X PRECEDING AND CURRENT ROW)
```

#### Top transacción de cada cliente en los ultimos 3 meses

````python
# fecha más reciente en los datos
sparksession.sql("""
SELECT MAX(Transaction_Date) AS Latest_Date
FROM BodoqueSQL
""").show()
````

Vemos que la ultima compra fue el 04 de marzo de este año, por lo que veremos los 3 ultimos meses desde esa fecha

````python
#Consulta SQL para calcular la compra más alta en los últimos tres meses
query = """
WITH RecentPurchases AS (
    SELECT
        Customer_ID,
        Transaction_Date,
        Transaction_Amount
    FROM BodoqueSQL
    WHERE Transaction_Date BETWEEN TIMESTAMP('2024-01-03 00:00:00') AND TIMESTAMP('2024-04-03 23:59:59')
),
RankedPurchases AS (
    SELECT
        Customer_ID,
        Transaction_Date,
        Transaction_Amount,
        ROW_NUMBER() OVER (PARTITION BY Customer_ID ORDER BY Transaction_Amount DESC) AS Rank
    FROM RecentPurchases
)
SELECT
    Customer_ID,
    Transaction_Date,
    Transaction_Amount
FROM RankedPurchases
WHERE Rank = 1
ORDER BY Transaction_Amount DESC

"""

result_df = sparksession.sql(query)
result_df.show()
````


En PySpark, las funciones de ventana se utilizan para realizar cálculos sobre un grupo de datos dentro de una partición específica sin agrupar completamente los datos. Esto es útil cuando necesitas aplicar funciones agregadas o de ranking a los datos mientras mantienes el detalle de cada fila.

Para calcular la compra más alta realizada por cada usuario en los últimos tres meses en PySpark, primero debes filtrar los datos para el rango de fechas deseado utilizando operaciones de filtrado. Luego, puedes definir una ventana usando Window.partitionBy() para dividir los datos en particiones basadas en el ID del cliente. Dentro de cada partición, puedes ordenar las filas por el monto de la transacción en orden descendente utilizando Window.orderBy(). Finalmente, se aplica una función de ranking, como row_number(), para asignar un número a cada fila dentro de la partición. La fila con el número 1 representa la transacción con el monto más alto. Esta metodología permite identificar la compra más alta de cada cliente mientras se mantiene la información de todas las transacciones individuales

Las UDFs ofrecen flexibilidad y personalización al permitir la implementación de lógica específica que no está disponible en las funciones integradas de Spark. Esto es útil cuando se requiere realizar cálculos complejos o aplicar modelos personalizados que no se pueden expresar con funciones predefinidas. Sin embargo, las UDFs pueden ser menos eficientes en comparación con las funciones integradas o SQL, ya que no siempre se benefician del optimizador Catalyst de Spark. Catalyst optimiza consultas SQL y funciones integradas para mejorar el rendimiento mediante técnicas como la reescritura de consultas y la ejecución en paralelo. Las UDFs, por otro lado, pueden introducir una sobrecarga adicional debido a la serialización y deserialización de datos, y su ejecución personalizada puede no estar tan bien optimizada como las operaciones nativas.
